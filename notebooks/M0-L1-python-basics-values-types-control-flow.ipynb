## Numbers, strings, and variables

A variable is a name attached to a value; Python infers the **type**. `42` is an `int`, `3.14` a `float`, `"cat"` a `str`, `True` a `bool`. Check with `type(x)`; convert with `int()`, `float()`, `str()`.



```python
acc = 0.875              # float — model scores live here
n = 3                  # int — counts and indices
label = "cat"          # str — labels, filenames, column names
ok = acc > 0.8         # bool — every comparison produces one

int("42")              # 42          (int("cat") would raise ValueError)
type(acc)              # <class 'float'>
```

Two operators to memorize: `//` divides and rounds **down** to a whole number, `%` gives the remainder. Together they answer "how many full groups of 32 fit in 1000 samples, and how many are left over":



```python
1000 // 32     # 31 full groups
1000 % 32      # 8 samples left over
2 ** 10        # 1024 — ** is exponentiation, not ^
9 / 2          # 4.5  — / always returns a float
-7 // 2        # -4   — floor division rounds DOWN, not toward 0
```

- `==` compares values, `is` compares identity — for numbers and strings always use `==`.
- `0.1 + 0.2 == 0.3` is `False`: floats are binary fractions, not exact decimals. Compare with `abs(a - b) < 1e-9`.
- `True` counts as `1`, so `sum(mask)` — where `mask` is a list of True/False flags — counts the Trues; that one trick computes accuracies, filtered lengths, and selection coverage all over this course.



## Strings and f-strings

Strings are immutable sequences of characters — you never change `"cat"` into `"car"`, you build a new string. The daily tools: `split` (text → list), `join` (list → text), `strip` (removes whitespace and newlines at both ends), `replace`, and `in` for substring tests.



```python
line = "cat,dog,bird"
parts = line.split(",")          # ['cat', 'dog', 'bird']
";".join(parts)                  # 'cat;dog;bird' — glue first, then .join
"  padded\n".strip()             # 'padded'
"cat" in "concatenate"           # True — substring test
```

**f-strings** evaluate whatever sits in `{}` and insert it into the text. `:.3f` formats a float with 3 decimals — the standard way to print a score:



```python
loss = 0.45678
f"epoch {3}, loss={loss:.4f}"    # 'epoch 3, loss=0.4568'
f"{0.8765:.1%}"                  # '87.6%' — percent formatting
f"{'cat':>8}"                    # '     cat' — right-align for neat logs
```

Indexing follows the list rules coming next: `s[0]` is the first character, `s[-1]` the last, `s[1:4]` the slice from index 1 up to (not including) 4, `len(s)` the length.



## Lists: indexing, slicing, mutating

A **list** is an ordered, mutable sequence — the default container. Indexing is 0-based: `xs[0]` first, `xs[-1]` last. Slices `xs[a:b]` take indices a through b−1 — the end is **exclusive**, so `len(xs[a:b])` is `b - a`.



```python
xs = [10, 20, 30, 40, 50]
xs[0], xs[-1]            # (10, 50)
xs[1:3]                  # [20, 30] — indices 1 and 2 only
xs[-2:]                  # [40, 50] — the last two
# full slice form is xs[start:stop:step] — step −1 walks backwards:
xs[::-1]                 # [50, 40, 30, 20, 10] — reversed

xs.append(60)            # grow by one: [10, 20, 30, 40, 50, 60]
xs.pop()                 # remove and return the last item
len(xs), 20 in xs        # (5, True)
```

`sorted(xs)` returns a **new** sorted list; `xs.sort()` sorts in place and returns `None`. `xs = xs.sort()` throws the list away — a classic silent bug. Same story for `xs.reverse()`.

> **Note.** Strings and tuples follow the same indexing and slicing rules but are immutable: `"abc"[0]` is `"a"`, while `"abc"[0] = "x"` is a TypeError.



## Conditions and truthiness

`if` / `elif` / `else` picks a branch; `and`, `or`, `not` combine tests; chained comparisons read like math: `0 <= p <= 1`.



```python
acc = 0.85
if acc >= 0.9:
    grade = "gold"
elif acc >= 0.7:
    grade = "silver"
else:
    grade = "keep training"      # acc is 0.85 → grade is 'silver'
```

Every value has a truth value. **Falsy**: `False`, `0`, `0.0`, `""`, `[]`, `{}`, `None`. Everything else is truthy, so `if scores:` reads as "if scores is non-empty".



```python
scores = []
if scores:                       # empty list → falsy → skips
    print(max(scores))
else:
    print("nothing yet")         # this prints

x = None
if x is None:                    # None is compared with "is", never ==
    print("missing")
```

- Indentation **is** the syntax: the block under `if` must be indented (4 spaces) — there are no braces.
- `if x:` treats a real `0` the same as "no value". When zero is legal, write `if x is not None:`.



## Loops: for, while, enumerate, zip

`for` iterates over items directly — no index bookkeeping. `range(n)` yields 0, 1, …, n−1; `enumerate(seq)` hands back (index, item) pairs; `zip(a, b)` walks two lists in lockstep and stops at the shorter one.



```python
scores = [0.71, 0.85, 0.64]
total = 0
for s in scores:                 # the items themselves, one by one
    total += s                   # shorthand for total = total + s
total / len(scores)              # 0.7333...

best, best_i = -1.0, -1
for i, s in enumerate(scores):
    if s > best:
        best, best_i = s, i      # two assignments at once — tuple unpacking
best_i                           # 1 — the argmax index, found by hand
```

```python
for name, i in zip(["cat", "dog"], [0, 1]):
    print(name, i)               # cat 0, then dog 1

n = 100
while n > 1:                     # repeats until the condition fails
    n //= 2                      # 50, 25, 12, 6, 3, 1
n                                # 1
```

- `break` exits the loop early (stop at the first corrupt row); `continue` skips to the next item.
- `for i in range(len(xs))` works but is un-Pythonic — `enumerate` gives index and value without indexing back in.
- Never append to or delete from a list while looping over it — collect into a new list instead.



## When it breaks: reading a traceback

> **Note.** Notebook words: a **cell** is one runnable block of a notebook; the **kernel** is the Python engine that runs the cells.

Errors print a **traceback** — a stack of calls ending with the actual error. Read it bottom-up: the last line names the error type and message, the lines above show your code. The four you will meet daily:

- `NameError`: a name that doesn't exist — a typo, or a cell you never ran.
- `TypeError`: wrong type for the operation — `"3" + 2`, or calling something that isn't a function.
- `IndexError` / `KeyError`: a list index past the end, or a lookup key that isn't stored.
- `SyntaxError` / `IndentationError`: Python can't parse the code — check missing colons and mixed indentation.



```python
xs = [1, 2, 3]
# xs[5]                  # IndexError: list index out of range
# xs + 5                 # TypeError: can only concatenate list to list
# pritn(xs)              # NameError: name 'pritn' is not defined
x = xs[2] + int("4")     # 7 — int() parses the string first
```

> **Note.** In a notebook the sneakiest trap is stale state: a cell can keep working on a variable you deleted or renamed an hour ago. When results look impossible, **restart the kernel and run all cells top to bottom**.



### ✏️ What does `7 // 2 + 7 % 2` equal?

**A.** 5
**B.** 4
**C.** 3.5
**D.** 3

<details><summary>Answer</summary>

**B.** 4

7 // 2 is 3 and 7 % 2 is 1, so the sum is 4. In general a // b * b + a % b == a.

</details>

### ✏️ xs = [10, 20, 30]. What is xs[1:]?

**A.** [10, 20, 30]
**B.** [20]
**C.** [20, 30]
**D.** [10]

<details><summary>Answer</summary>

**C.** [20, 30]

The slice starts at index 1 and runs to the end; omitting the end means "everything".

</details>

```python
# ── task-checking helper cell — run once ──
import copy
import numpy as np
import math, statistics, itertools, collections, functools, heapq, random

try:
    import sys; sys.stdout.reconfigure(encoding="utf-8", errors="replace")
except Exception:
    pass

_L = {"noFn":"❓ Define a function `{e}` in the solution cell and run it","runErr":"❌ Test {i}: {t}: {m}","wrong":"❌ Test {i}: {r}","badRef":"⚠️ Check harness broke (generator/reference): {t}: {m}","inL":"Input","expL":"Expected","gotL":"Got","ok":"✅ Correct! All {n} tests passed."}

def _nb_copy_kw(kw):
    return {k: (v.copy() if isinstance(v, np.ndarray) else copy.deepcopy(v)) for k, v in kw.items()}

def _nb_jsafe(x, depth=0):
    if depth > 6: return "…"
    if isinstance(x, np.ndarray): return _nb_jsafe(x.tolist(), depth + 1)
    if isinstance(x, np.integer): return int(x)
    if isinstance(x, np.floating): return float(x)
    if isinstance(x, np.bool_): return bool(x)
    if isinstance(x, float): return round(x, 6) if math.isfinite(x) else str(x)
    if isinstance(x, (list, tuple)): return [_nb_jsafe(v, depth + 1) for v in x]
    if isinstance(x, dict): return {str(k): _nb_jsafe(v, depth + 1) for k, v in x.items()}
    if isinstance(x, (int, str, bool)) or x is None: return x
    return repr(x)

def _nb_cmp(exp, got, path=""):
    if isinstance(exp, np.ndarray) or isinstance(got, np.ndarray):
        e, g = np.asarray(exp), np.asarray(got)
        if e.shape != g.shape:
            return f"shape {tuple(g.shape)} != expected {tuple(e.shape)}"
        try:
            if np.issubdtype(e.dtype, np.number) or np.issubdtype(g.dtype, np.number):
                ef, gf = e.astype(float), g.astype(float)
                close = np.isclose(gf, ef, atol=1e-6, rtol=1e-5, equal_nan=True)
                if not bool(close.all()):
                    i = tuple(int(v) for v in np.argwhere(~close)[0])
                    return f"{path}first mismatch at index {i}: got {gf[i]:.6g} vs expected {ef[i]:.6g}"
            elif not np.array_equal(g, e):
                return f"{path}values differ"
        except Exception as ex:
            return f"{path}cannot compare: {ex}"
        return None
    if isinstance(exp, (int, float)) and not isinstance(exp, bool) and isinstance(got, (int, float)) and not isinstance(got, bool):
        if not np.isclose(float(exp), float(got), atol=1e-6, rtol=1e-5):
            return f"{path}{got} != expected {exp}"
        return None
    if isinstance(exp, (list, tuple)) and isinstance(got, (list, tuple)):
        if len(exp) != len(got):
            return f"{path}length {len(got)} != expected {len(exp)}"
        for i, (a, b) in enumerate(zip(exp, got)):
            r = _nb_cmp(a, b, path + f"[{i}].")
            if r: return r
        return None
    if isinstance(exp, dict) and isinstance(got, dict):
        if set(exp) != set(got):
            return f"{path}keys differ"
        for k in exp:
            r = _nb_cmp(exp[k], got[k], path + f"[{k!r}].")
            if r: return r
        return None
    if exp != got:
        return f"{path}{got!r} != expected {exp!r}"
    return None

def _check_task(g, ref_src, entry, n=20, seed=0):
    """g — globals() ноутбука; ref_src — gen() + эталонное решение."""
    fn = g.get(entry)
    if not callable(fn):
        print(_L["noFn"].format(e=entry)); return
    ns = {"np": np, "math": math, "statistics": statistics, "itertools": itertools,
          "collections": collections, "functools": functools, "heapq": heapq, "random": random}
    try:
        exec(ref_src, ns)
        gen, ref = ns["gen"], ns[entry]
    except Exception as e:
        print(_L["badRef"].format(t=type(e).__name__, m=e)); return
    rng = np.random.default_rng(seed)
    for i in range(n):
        try:
            kw = gen(rng)
        except Exception as e:
            print(_L["badRef"].format(t=type(e).__name__, m=e)); return
        try:
            got = fn(**_nb_copy_kw(kw))
        except Exception as e:
            print(_L["runErr"].format(i=i + 1, t=type(e).__name__, m=e)
                  + "\n" + _L["inL"] + ": " + repr(_nb_jsafe(kw))); return
        exp = ref(**_nb_copy_kw(kw))
        r = _nb_cmp(exp, got)
        if r:
            print(_L["wrong"].format(i=i + 1, r=r)
                  + "\n" + _L["inL"] + ": " + repr(_nb_jsafe(kw))
                  + "\n" + _L["expL"] + ": " + repr(_nb_jsafe(exp))
                  + "\n" + _L["gotL"] + ": " + repr(_nb_jsafe(got))); return
    print(_L["ok"].format(n=n))
```
